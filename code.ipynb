```python
# ================================================================
# INTEGRATED FOOD-WASTE PREDICTION AND DECISION-SUPPORT FRAMEWORK
# ================================================================
#
# Research framework:
# Data Preparation
#       ↓
# Data Warehouse / OLAP-style Analysis
#       ↓
# Clustering
#       ↓
# Association Rule Mining
#       ↓
# Machine-Learning Regression
#       ↓
# Feature Ablation
#       ↓
# Outlier Sensitivity
#       ↓
# Time-Aware Model Selection
#       ↓
# Final Future-Test Evaluation
#       ↓
# SHAP Explainability
#       ↓
# Decision Support
#
# IMPORTANT:
# - The final test period is never used for model selection.
# - Splitting is performed using unique dates to prevent same-date leakage.
# - Model selection uses mean RMSE from chronological validation.
# - Random Forest is NOT assumed to be the winner.
# - The model with the lowest validation RMSE is selected automatically.
# ================================================================


# ================================================================
# 1. INSTALL REQUIRED PACKAGES
# ================================================================

!pip -q install xgboost mlxtend shap openpyxl


# ================================================================
# 2. IMPORT LIBRARIES
# ================================================================

import os
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder, StandardScaler, RobustScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

from sklearn.impute import SimpleImputer

from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from xgboost import XGBRegressor

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score,
    silhouette_score
)

from sklearn.cluster import KMeans

from mlxtend.frequent_patterns import apriori, association_rules

import shap


# ================================================================
# 3. LOAD DATA
# ================================================================

FILE_PATH = "food_waste_cleaned.csv"

if not os.path.exists(FILE_PATH):
    raise FileNotFoundError(
        f"{FILE_PATH} was not found. "
        "Upload food_waste_cleaned.csv to Google Colab first."
    )

df = pd.read_csv(FILE_PATH)

print("=" * 70)
print("DATASET LOADED")
print("=" * 70)

print("Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())


# ================================================================
# 4. BASIC DATA CLEANING
# ================================================================

df = df.copy()

# Standardize column names
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

# Required columns
required_columns = [
    "date",
    "meals_served",
    "kitchen_staff",
    "temperature_c",
    "humidity_percent",
    "day_of_week",
    "special_event",
    "past_waste_kg",
    "food_waste_kg"
]

missing_required = [
    col for col in required_columns
    if col not in df.columns
]

if missing_required:
    raise ValueError(
        f"Missing required columns: {missing_required}"
    )

# Convert date
df["date"] = pd.to_datetime(
    df["date"],
    errors="coerce"
)

# Remove rows with invalid dates
df = df.dropna(subset=["date"]).copy()

# Remove exact duplicate rows
before_duplicates = len(df)

df = df.drop_duplicates().copy()

after_duplicates = len(df)

print("\nDuplicate rows removed:",
      before_duplicates - after_duplicates)


# ================================================================
# 5. NUMERIC CONVERSION
# ================================================================

numeric_columns = [
    "meals_served",
    "kitchen_staff",
    "temperature_c",
    "humidity_percent",
    "day_of_week",
    "special_event",
    "past_waste_kg",
    "food_waste_kg"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )

# Remove rows where target is missing
df = df.dropna(
    subset=["food_waste_kg"]
).copy()


# ================================================================
# 6. TEMPORAL FEATURE ENGINEERING
# ================================================================

df["year"] = df["date"].dt.year
df["month"] = df["date"].dt.month
df["day"] = df["date"].dt.day
df["day_of_week_num"] = df["date"].dt.dayofweek
df["week_of_year"] = df["date"].dt.isocalendar().week.astype(int)
df["is_weekend"] = (
    df["day_of_week_num"] >= 5
).astype(int)

# Waste per meal is a descriptive/decision-support metric.
# It is NOT used as a predictive feature because it contains
# the target variable food_waste_kg.
df["waste_per_meal"] = (
    df["food_waste_kg"] /
    df["meals_served"].replace(0, np.nan)
)

df["waste_per_100_meals"] = (
    df["waste_per_meal"] * 100
)


# ================================================================
# 7. FINAL DATA QUALITY CHECK
# ================================================================

print("\n" + "=" * 70)
print("DATA QUALITY")
print("=" * 70)

print("Final shape:", df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nNumber of unique dates:",
      df["date"].nunique())

print("Date range:",
      df["date"].min().date(),
      "to",
      df["date"].max().date())


# ================================================================
# 8. TABLE 1 — DATASET DESCRIPTION
# ================================================================

dataset_description = pd.DataFrame({
    "Variable": [
        "date",
        "meals_served",
        "kitchen_staff",
        "temperature_C",
        "humidity_percent",
        "day_of_week",
        "special_event",
        "past_waste_kg",
        "food_waste_kg"
    ],
    "Type": [
        "Date",
        "Numerical",
        "Numerical",
        "Numerical",
        "Numerical",
        "Categorical/Numerical",
        "Binary",
        "Numerical",
        "Numerical"
    ],
    "Role": [
        "Temporal",
        "Predictor",
        "Predictor",
        "Predictor",
        "Predictor",
        "Predictor",
        "Predictor",
        "Predictor",
        "Target"
    ],
    "Description": [
        "Observation date",
        "Number of meals served",
        "Number of kitchen staff",
        "Temperature in Celsius",
        "Relative humidity percentage",
        "Day-of-week indicator",
        "Special-event indicator",
        "Historical food waste",
        "Food waste to be predicted"
    ]
})

print("\nTABLE 1 — DATASET DESCRIPTION")
display(dataset_description)


# ================================================================
# 9. TABLE 2 — DESCRIPTIVE STATISTICS
# ================================================================

descriptive_columns = [
    "meals_served",
    "kitchen_staff",
    "temperature_c",
    "humidity_percent",
    "past_waste_kg",
    "food_waste_kg",
    "waste_per_meal"
]

descriptive_stats = df[
    descriptive_columns
].describe().T.reset_index()

descriptive_stats = descriptive_stats.rename(
    columns={
        "index": "Variable",
        "count": "N",
        "mean": "Mean",
        "std": "Std_Dev",
        "min": "Minimum",
        "25%": "Q1",
        "50%": "Median",
        "75%": "Q3",
        "max": "Maximum"
    }
)

print("\nTABLE 2 — DESCRIPTIVE STATISTICS")
display(descriptive_stats)


# ================================================================
# 10. FOOD-WASTE DISTRIBUTION
# ================================================================

plt.figure(figsize=(9, 5))

sns.histplot(
    df["food_waste_kg"],
    kde=True
)

plt.title("Distribution of Food Waste")
plt.xlabel("Food Waste (kg)")
plt.ylabel("Frequency")
plt.tight_layout()

plt.savefig(
    "figure_1_food_waste_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 11. OLAP-STYLE ANALYSIS
# ================================================================

# Monthly waste
monthly_olap = (
    df.groupby(
        ["year", "month"],
        as_index=False
    )
    .agg(
        total_food_waste_kg=("food_waste_kg", "sum"),
        average_food_waste_kg=("food_waste_kg", "mean"),
        total_meals_served=("meals_served", "sum"),
        average_waste_per_meal=("waste_per_meal", "mean")
    )
)

# Event analysis
event_olap = (
    df.groupby(
        "special_event",
        as_index=False
    )
    .agg(
        observations=("food_waste_kg", "size"),
        average_food_waste_kg=("food_waste_kg", "mean"),
        average_meals_served=("meals_served", "mean")
    )
)

# Staff analysis
staff_olap = (
    df.groupby(
        "kitchen_staff",
        as_index=False
    )
    .agg(
        observations=("food_waste_kg", "size"),
        average_food_waste_kg=("food_waste_kg", "mean"),
        average_meals_served=("meals_served", "mean")
    )
)

print("\nOLAP — MONTHLY")
display(monthly_olap.head(20))

print("\nOLAP — SPECIAL EVENT")
display(event_olap)

print("\nOLAP — KITCHEN STAFF")
display(staff_olap)


# ================================================================
# 12. CLUSTERING
# ================================================================

cluster_features = [
    "meals_served",
    "kitchen_staff",
    "temperature_c",
    "humidity_percent",
    "past_waste_kg",
    "food_waste_kg"
]

cluster_df = df[
    cluster_features
].dropna().copy()

# Robust scaling because the dataset contains extreme observations
cluster_scaler = RobustScaler()

X_cluster = cluster_scaler.fit_transform(
    cluster_df
)

cluster_scores = []

for k in range(2, 7):

    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=20
    )

    labels = kmeans.fit_predict(
        X_cluster
    )

    score = silhouette_score(
        X_cluster,
        labels
    )

    cluster_scores.append({
        "K": k,
        "Silhouette_Score": score
    })

cluster_scores_df = pd.DataFrame(
    cluster_scores
)

best_k = int(
    cluster_scores_df.loc[
        cluster_scores_df["Silhouette_Score"].idxmax(),
        "K"
    ]
)

print("\nCLUSTERING EVALUATION")
display(cluster_scores_df)

print("Selected number of clusters:", best_k)


# Final clustering
final_kmeans = KMeans(
    n_clusters=best_k,
    random_state=42,
    n_init=20
)

cluster_df["cluster"] = (
    final_kmeans.fit_predict(
        X_cluster
    )
)

cluster_profiles = (
    cluster_df
    .groupby("cluster")
    .agg(
        observations=("food_waste_kg", "size"),
        avg_meals_served=("meals_served", "mean"),
        avg_kitchen_staff=("kitchen_staff", "mean"),
        avg_temperature_C=("temperature_c", "mean"),
        avg_humidity_percent=("humidity_percent", "mean"),
        avg_past_waste_kg=("past_waste_kg", "mean"),
        avg_food_waste_kg=("food_waste_kg", "mean")
    )
    .reset_index()
)

print("\nCLUSTER PROFILES")
display(cluster_profiles)


# ================================================================
# 13. ASSOCIATION-RULE MINING
# ================================================================

association_df = df[
    [
        "meals_served",
        "kitchen_staff",
        "past_waste_kg",
        "food_waste_kg",
        "special_event"
    ]
].copy()

# Convert continuous variables into meaningful categories
association_df["meals_high"] = (
    association_df["meals_served"]
    >= association_df["meals_served"].median()
)

association_df["staff_high"] = (
    association_df["kitchen_staff"]
    >= association_df["kitchen_staff"].median()
)

association_df["past_waste_high"] = (
    association_df["past_waste_kg"]
    >= association_df["past_waste_kg"].median()
)

association_df["waste_high"] = (
    association_df["food_waste_kg"]
    >= association_df["food_waste_kg"].median()
)

association_df["special_event_yes"] = (
    association_df["special_event"] == 1
)

transaction_columns = [
    "meals_high",
    "staff_high",
    "past_waste_high",
    "waste_high",
    "special_event_yes"
]

transactions = (
    association_df[
        transaction_columns
    ]
    .astype(bool)
)

frequent_itemsets = apriori(
    transactions,
    min_support=0.05,
    use_colnames=True
)

if len(frequent_itemsets) > 0:

    rules = association_rules(
        frequent_itemsets,
        metric="confidence",
        min_threshold=0.50
    )

    if len(rules) > 0:

        rules["antecedent_length"] = (
            rules["antecedents"].apply(len)
        )

        rules["consequent_length"] = (
            rules["consequents"].apply(len)
        )

        # Keep rules where high waste is the consequent
        high_waste_rules = rules[
            rules["consequents"].apply(
                lambda x: "waste_high" in x
            )
        ].copy()

        high_waste_rules = (
            high_waste_rules
            .sort_values(
                ["lift", "confidence"],
                ascending=False
            )
            .reset_index(drop=True)
        )

        # Convert frozensets to readable strings
        high_waste_rules["Antecedent"] = (
            high_waste_rules["antecedents"]
            .apply(lambda x: ", ".join(sorted(x)))
        )

        high_waste_rules["Consequent"] = (
            high_waste_rules["consequents"]
            .apply(lambda x: ", ".join(sorted(x)))
        )

        association_table = high_waste_rules[
            [
                "Antecedent",
                "Consequent",
                "support",
                "confidence",
                "lift"
            ]
        ].copy()

        association_table.columns = [
            "Antecedent",
            "Consequent",
            "Support",
            "Confidence",
            "Lift"
        ]

    else:
        association_table = pd.DataFrame(
            columns=[
                "Antecedent",
                "Consequent",
                "Support",
                "Confidence",
                "Lift"
            ]
        )

else:

    association_table = pd.DataFrame(
        columns=[
            "Antecedent",
            "Consequent",
            "Support",
            "Confidence",
            "Lift"
        ]
    )

print("\nTABLE 3 — ASSOCIATION RULES")
display(association_table.head(15))


# ================================================================
# 14. PREDICTIVE MODELING DATA
# ================================================================

# IMPORTANT:
# Do NOT use:
# - food_waste_kg as a feature
# - waste_per_meal as a feature
# - waste_per_100_meals as a feature
#
# because they contain the target information.

feature_columns = [
    "meals_served",
    "kitchen_staff",
    "temperature_c",
    "humidity_percent",
    "day_of_week",
    "special_event",
    "past_waste_kg",
    "year",
    "month",
    "day",
    "day_of_week_num",
    "week_of_year",
    "is_weekend"
]

target_column = "food_waste_kg"

model_df = df[
    ["date"] +
    feature_columns +
    [target_column]
].copy()

model_df = model_df.sort_values(
    "date"
).reset_index(drop=True)


# ================================================================
# 15. CHRONOLOGICAL 80/20 SPLIT BY UNIQUE DATES
# ================================================================

unique_dates = (
    model_df["date"]
    .drop_duplicates()
    .sort_values()
    .to_numpy()
)

split_date_index = int(
    len(unique_dates) * 0.80
)

development_dates = unique_dates[
    :split_date_index
]

final_test_dates = unique_dates[
    split_date_index:
]

development_df = model_df[
    model_df["date"].isin(
        development_dates
    )
].copy()

final_test_df = model_df[
    model_df["date"].isin(
        final_test_dates
    )
].copy()

# Verify no date leakage
date_overlap = set(
    development_df["date"]
).intersection(
    set(final_test_df["date"])
)

if len(date_overlap) > 0:
    raise ValueError(
        "Date leakage detected between development and final test sets."
    )

print("\n" + "=" * 70)
print("CHRONOLOGICAL SPLIT")
print("=" * 70)

print(
    "Development observations:",
    len(development_df)
)

print(
    "Final test observations:",
    len(final_test_df)
)

print(
    "Development date range:",
    development_df["date"].min().date(),
    "to",
    development_df["date"].max().date()
)

print(
    "Final test date range:",
    final_test_df["date"].min().date(),
    "to",
    final_test_df["date"].max().date()
)

print(
    "Shared dates:",
    len(date_overlap)
)


# ================================================================
# 16. PREPROCESSOR
# ================================================================

categorical_features = [
    "day_of_week",
    "special_event",
    "month",
    "day_of_week_num",
    "is_weekend"
]

numeric_features = [
    col for col in feature_columns
    if col not in categorical_features
]

numeric_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="median")
        )
    ]
)

categorical_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),
        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ]
)

preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            numeric_transformer,
            numeric_features
        ),
        (
            "cat",
            categorical_transformer,
            categorical_features
        )
    ],
    remainder="drop"
)


# ================================================================
# 17. MODEL DEFINITIONS
# ================================================================

models = {

    "Linear Regression":
        LinearRegression(),

    "Random Forest":
        RandomForestRegressor(
            n_estimators=300,
            random_state=42,
            n_jobs=-1
        ),

    "Gradient Boosting":
        GradientBoostingRegressor(
            n_estimators=300,
            learning_rate=0.05,
            max_depth=3,
            random_state=42
        ),

    "XGBoost":
        XGBRegressor(
            n_estimators=300,
            learning_rate=0.05,
            max_depth=4,
            subsample=0.8,
            colsample_bytree=0.8,
            objective="reg:squarederror",
            random_state=42,
            n_jobs=-1
        )
}


# ================================================================
# 18. CHRONOLOGICAL VALIDATION FOLDS
# ================================================================
#
# Validation is performed ONLY inside the development period.
#
# Each fold:
# training data = earlier dates
# validation data = immediately later dates
#
# This avoids future information entering the training data.
# ================================================================

development_unique_dates = (
    development_df["date"]
    .drop_duplicates()
    .sort_values()
    .to_numpy()
)

n_dates = len(
    development_unique_dates
)

# Three expanding-window validation folds
fold_boundaries = [
    (0.50, 0.65),
    (0.65, 0.80),
    (0.80, 1.00)
]

validation_results = []

for fold_number, (
    train_end_pct,
    valid_end_pct
) in enumerate(
    fold_boundaries,
    start=1
):

    train_end_idx = int(
        n_dates * train_end_pct
    )

    valid_end_idx = int(
        n_dates * valid_end_pct
    )

    train_dates = (
        development_unique_dates[
            :train_end_idx
        ]
    )

    valid_dates = (
        development_unique_dates[
            train_end_idx:valid_end_idx
        ]
    )

    train_fold = development_df[
        development_df["date"].isin(
            train_dates
        )
    ].copy()

    valid_fold = development_df[
        development_df["date"].isin(
            valid_dates
        )
    ].copy()

    X_train = train_fold[
        feature_columns
    ]

    y_train = train_fold[
        target_column
    ]

    X_valid = valid_fold[
        feature_columns
    ]

    y_valid = valid_fold[
        target_column
    ]

    for model_name, model in models.items():

        pipeline = Pipeline(
            steps=[
                (
                    "preprocessor",
                    preprocessor
                ),
                (
                    "model",
                    model
                )
            ]
        )

        pipeline.fit(
            X_train,
            y_train
        )

        predictions = pipeline.predict(
            X_valid
        )

        mae = mean_absolute_error(
            y_valid,
            predictions
        )

        rmse = np.sqrt(
            mean_squared_error(
                y_valid,
                predictions
            )
        )

        r2 = r2_score(
            y_valid,
            predictions
        )

        validation_results.append({
            "Fold": fold_number,
            "Model": model_name,
            "MAE": mae,
            "RMSE": rmse,
            "R2": r2
        })


validation_results_df = pd.DataFrame(
    validation_results
)

print("\nTIME-AWARE VALIDATION RESULTS")
display(validation_results_df)


# ================================================================
# 19. TABLE 4 — MODEL COMPARISON
# ================================================================

model_comparison = (
    validation_results_df
    .groupby("Model")
    .agg(
        Mean_MAE=("MAE", "mean"),
        Mean_RMSE=("RMSE", "mean"),
        Mean_R2=("R2", "mean")
    )
    .reset_index()
)

# Primary selection criterion:
# lowest mean RMSE.
#
# MAE and R2 are secondary reporting metrics.
model_comparison = (
    model_comparison
    .sort_values(
        [
            "Mean_RMSE",
            "Mean_MAE"
        ],
        ascending=[
            True,
            True
        ]
    )
    .reset_index(drop=True)
)

print("\nTABLE 4 — TIME-AWARE MODEL COMPARISON")
display(model_comparison)


# ================================================================
# 20. SELECT FINAL MODEL
# ================================================================

selected_model_name = (
    model_comparison
    .iloc[0]["Model"]
)

selected_model = models[
    selected_model_name
]

print(
    "\nSELECTED FINAL MODEL:",
    selected_model_name
)

print(
    "Selection criterion: lowest mean RMSE "
    "across chronological validation folds."
)


# ================================================================
# 21. TABLE 5 — FEATURE ABLATION
# ================================================================

ablation_feature_sets = {

    "All Features": feature_columns,

    "Meals + Past Waste": [
        "meals_served",
        "past_waste_kg"
    ],

    "Meals Only": [
        "meals_served"
    ],

    "All Except Meals": [
        col for col in feature_columns
        if col != "meals_served"
    ]
}

# Use the SAME chronological development/test design
# for a fair comparison.
#
# We use the final validation period of the development set
# as the ablation evaluation period.

ablation_train_dates = development_unique_dates[
    :int(n_dates * 0.80)
]

ablation_valid_dates = development_unique_dates[
    int(n_dates * 0.80):
]

ablation_train = development_df[
    development_df["date"].isin(
        ablation_train_dates
    )
].copy()

ablation_valid = development_df[
    development_df["date"].isin(
        ablation_valid_dates
    )
].copy()

ablation_results = []

for feature_set_name, features in (
    ablation_feature_sets.items()
):

    X_train_ab = ablation_train[
        features
    ]

    y_train_ab = ablation_train[
        target_column
    ]

    X_valid_ab = ablation_valid[
        features
    ]

    y_valid_ab = ablation_valid[
        target_column
    ]

    # Automatically determine categorical/numeric columns
    cat_ab = [
        c for c in features
        if c in categorical_features
    ]

    num_ab = [
        c for c in features
        if c not in cat_ab
    ]

    preprocessor_ab = ColumnTransformer(
        transformers=[
            (
                "num",
                Pipeline(
                    steps=[
                        (
                            "imputer",
                            SimpleImputer(
                                strategy="median"
                            )
                        )
                    ]
                ),
                num_ab
            ),
            (
                "cat",
                Pipeline(
                    steps=[
                        (
                            "imputer",
                            SimpleImputer(
                                strategy="most_frequent"
                            )
                        ),
                        (
                            "onehot",
                            OneHotEncoder(
                                handle_unknown="ignore",
                                sparse_output=False
                            )
                        )
                    ]
                ),
                cat_ab
            )
        ],
        remainder="drop"
    )

    # Use the selected final model
    ablation_model = Pipeline(
        steps=[
            (
                "preprocessor",
                preprocessor_ab
            ),
            (
                "model",
                models[selected_model_name]
            )
        ]
    )

    ablation_model.fit(
        X_train_ab,
        y_train_ab
    )

    ablation_predictions = (
        ablation_model.predict(
            X_valid_ab
        )
    )

    ablation_mae = mean_absolute_error(
        y_valid_ab,
        ablation_predictions
    )

    ablation_rmse = np.sqrt(
        mean_squared_error(
            y_valid_ab,
            ablation_predictions
        )
    )

    ablation_r2 = r2_score(
        y_valid_ab,
        ablation_predictions
    )

    ablation_results.append({
        "Feature_Set": feature_set_name,
        "MAE": ablation_mae,
        "RMSE": ablation_rmse,
        "R2": ablation_r2
    })


ablation_results_df = pd.DataFrame(
    ablation_results
)

print("\nTABLE 5 — FEATURE ABLATION")
display(ablation_results_df)


# ================================================================
# 22. OUTLIER / EXTREME-OBSERVATION SENSITIVITY
# ================================================================
#
# IMPORTANT:
# These are statistical extreme observations.
# They are NOT automatically considered erroneous.
#
# IQR thresholds are calculated using TRAINING DATA ONLY.
# ================================================================

outlier_train = ablation_train.copy()
outlier_valid = ablation_valid.copy()

outlier_numeric_columns = [
    "meals_served",
    "kitchen_staff",
    "temperature_c",
    "humidity_percent",
    "past_waste_kg"
]

outlier_mask = pd.Series(
    False,
    index=outlier_train.index
)

outlier_thresholds = []

for col in outlier_numeric_columns:

    q1 = outlier_train[
        col
    ].quantile(0.25)

    q3 = outlier_train[
        col
    ].quantile(0.75)

    iqr = q3 - q1

    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    current_mask = (
        (outlier_train[col] < lower) |
        (outlier_train[col] > upper)
    )

    outlier_mask = (
        outlier_mask |
        current_mask
    )

    outlier_thresholds.append({
        "Feature": col,
        "Q1": q1,
        "Q3": q3,
        "IQR": iqr,
        "Lower_Bound": lower,
        "Upper_Bound": upper
    })

outlier_thresholds_df = pd.DataFrame(
    outlier_thresholds
)

clean_training = outlier_train[
    ~outlier_mask
].copy()

print("\nStatistical extreme observations removed:",
      outlier_mask.sum())

print(
    "Training observations before:",
    len(outlier_train)
)

print(
    "Training observations after:",
    len(clean_training)
)


# Original model
original_outlier_model = Pipeline(
    steps=[
        (
            "preprocessor",
            preprocessor
        ),
        (
            "model",
            models[selected_model_name]
        )
    ]
)

original_outlier_model.fit(
    outlier_train[feature_columns],
    outlier_train[target_column]
)

original_predictions = (
    original_outlier_model.predict(
        outlier_valid[feature_columns]
    )
)

original_mae = mean_absolute_error(
    outlier_valid[target_column],
    original_predictions
)

original_rmse = np.sqrt(
    mean_squared_error(
        outlier_valid[target_column],
        original_predictions
    )
)

original_r2 = r2_score(
    outlier_valid[target_column],
    original_predictions
)


# Outlier-treated model
outlier_treated_model = Pipeline(
    steps=[
        (
            "preprocessor",
            preprocessor
        ),
        (
            "model",
            models[selected_model_name]
        )
    ]
)

outlier_treated_model.fit(
    clean_training[feature_columns],
    clean_training[target_column]
)

treated_predictions = (
    outlier_treated_model.predict(
        outlier_valid[feature_columns]
    )
)

treated_mae = mean_absolute_error(
    outlier_valid[target_column],
    treated_predictions
)

treated_rmse = np.sqrt(
    mean_squared_error(
        outlier_valid[target_column],
        treated_predictions
    )
)

treated_r2 = r2_score(
    outlier_valid[target_column],
    treated_predictions
)

outlier_results_df = pd.DataFrame({

    "Condition": [
        "Original Training Data",
        "Statistical Extreme Observations Removed"
    ],

    "MAE": [
        original_mae,
        treated_mae
    ],

    "RMSE": [
        original_rmse,
        treated_rmse
    ],

    "R2": [
        original_r2,
        treated_r2
    ],

    "Training_Observations": [
        len(outlier_train),
        len(clean_training)
    ]
})

print("\nTABLE 6 — OUTLIER SENSITIVITY")
display(outlier_results_df)


# ================================================================
# 23. TRAIN FINAL MODEL ON ALL DEVELOPMENT DATA
# ================================================================

X_development = development_df[
    feature_columns
]

y_development = development_df[
    target_column
]

X_final_test = final_test_df[
    feature_columns
]

y_final_test = final_test_df[
    target_column
]

final_pipeline = Pipeline(
    steps=[
        (
            "preprocessor",
            preprocessor
        ),
        (
            "model",
            models[selected_model_name]
        )
    ]
)

final_pipeline.fit(
    X_development,
    y_development
)


# ================================================================
# 24. FINAL UNTOUCHED FUTURE TEST
# ================================================================

final_predictions = (
    final_pipeline.predict(
        X_final_test
    )
)

final_mae = mean_absolute_error(
    y_final_test,
    final_predictions
)

final_rmse = np.sqrt(
    mean_squared_error(
        y_final_test,
        final_predictions
    )
)

final_r2 = r2_score(
    y_final_test,
    final_predictions
)

final_test_results = pd.DataFrame({
    "Final_Model": [
        selected_model_name
    ],
    "Test_MAE": [
        final_mae
    ],
    "Test_RMSE": [
        final_rmse
    ],
    "Test_R2": [
        final_r2
    ],
    "Development_Observations": [
        len(development_df)
    ],
    "Final_Test_Observations": [
        len(final_test_df)
    ]
})

print("\nTABLE 7 — FINAL MODEL TEST RESULTS")
display(final_test_results)


# ================================================================
# 25. ACTUAL VS PREDICTED
# ================================================================

prediction_results = final_test_df[
    ["date", "food_waste_kg"]
].copy()

prediction_results[
    "predicted_food_waste_kg"
] = final_predictions

prediction_results[
    "absolute_error_kg"
] = (
    prediction_results[
        "food_waste_kg"
    ] -
    prediction_results[
        "predicted_food_waste_kg"
    ]
).abs()

prediction_results[
    "percentage_error"
] = (
    prediction_results[
        "absolute_error_kg"
    ] /
    prediction_results[
        "food_waste_kg"
    ].replace(0, np.nan)
) * 100

print("\nFINAL TEST PREDICTIONS")
display(
    prediction_results.head(20)
)


# ================================================================
# 26. ACTUAL VS PREDICTED FIGURE
# ================================================================

plt.figure(figsize=(8, 6))

plt.scatter(
    y_final_test,
    final_predictions,
    alpha=0.7
)

min_value = min(
    y_final_test.min(),
    final_predictions.min()
)

max_value = max(
    y_final_test.max(),
    final_predictions.max()
)

plt.plot(
    [min_value, max_value],
    [min_value, max_value],
    linestyle="--"
)

plt.xlabel("Actual Food Waste (kg)")
plt.ylabel("Predicted Food Waste (kg)")
plt.title(
    f"Actual vs Predicted Food Waste — {selected_model_name}"
)

plt.tight_layout()

plt.savefig(
    "figure_2_actual_vs_predicted.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 27. RESIDUAL ANALYSIS
# ================================================================

residuals = (
    y_final_test.to_numpy()
    -
    final_predictions
)

plt.figure(figsize=(9, 5))

plt.scatter(
    final_predictions,
    residuals,
    alpha=0.7
)

plt.axhline(
    0,
    linestyle="--"
)

plt.xlabel("Predicted Food Waste (kg)")
plt.ylabel("Residual (kg)")
plt.title("Residual Analysis")

plt.tight_layout()

plt.savefig(
    "figure_3_residual_analysis.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 28. SHAP EXPLAINABILITY
# ================================================================
#
# Model-agnostic SHAP is used so that the code remains valid
# regardless of which model wins time-aware validation.
#
# A limited background/sample size is used for computational
# efficiency in Google Colab.
# ================================================================

shap_background_size = min(
    100,
    len(X_development)
)

shap_sample_size = min(
    200,
    len(X_final_test)
)

shap_background = (
    X_development
    .sample(
        n=shap_background_size,
        random_state=42
    )
    .copy()
)

shap_sample = (
    X_final_test
    .sample(
        n=shap_sample_size,
        random_state=42
    )
    .copy()
)

# Model-agnostic SHAP explainer
explainer = shap.Explainer(
    final_pipeline.predict,
    shap_background
)

shap_values = explainer(
    shap_sample
)

# Mean absolute SHAP importance
shap_importance = pd.DataFrame({
    "Feature": shap_sample.columns,
    "Mean_Absolute_SHAP": np.abs(
        shap_values.values
    ).mean(axis=0)
})

shap_importance = (
    shap_importance
    .sort_values(
        "Mean_Absolute_SHAP",
        ascending=False
    )
    .reset_index(drop=True)
)

shap_importance[
    "Rank"
] = np.arange(
    1,
    len(shap_importance) + 1
)

shap_importance = shap_importance[
    [
        "Rank",
        "Feature",
        "Mean_Absolute_SHAP"
    ]
]

print("\nTABLE 8 — SHAP FEATURE IMPORTANCE")
display(shap_importance)


# ================================================================
# 29. SHAP SUMMARY PLOT
# ================================================================

plt.figure()

shap.summary_plot(
    shap_values,
    shap_sample,
    show=False
)

plt.tight_layout()

plt.savefig(
    "figure_4_shap_summary.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 30. DECISION-SUPPORT RISK CATEGORIES
# ================================================================
#
# Risk categories are based on the TRAINING target distribution,
# not the final test data, to avoid using test information.
#
# Low    <= 50th percentile
# Medium > 50th and <= 75th percentile
# High   > 75th percentile
# ================================================================

q50 = y_development.quantile(0.50)
q75 = y_development.quantile(0.75)

decision_support = prediction_results.copy()

decision_support[
    "risk_category"
] = pd.cut(
    decision_support[
        "predicted_food_waste_kg"
    ],
    bins=[
        -np.inf,
        q50,
        q75,
        np.inf
    ],
    labels=[
        "Low",
        "Medium",
        "High"
    ]
)

decision_support[
    "predicted_waste_per_meal"
] = (
    decision_support[
        "predicted_food_waste_kg"
    ] /
    final_test_df[
        "meals_served"
    ].to_numpy()
)

decision_support[
    "predicted_waste_per_100_meals"
] = (
    decision_support[
        "predicted_waste_per_meal"
    ] * 100
)

print("\nTABLE 9 — DECISION-SUPPORT PREDICTIONS")
display(
    decision_support[
        [
            "date",
            "food_waste_kg",
            "predicted_food_waste_kg",
            "absolute_error_kg",
            "risk_category",
            "predicted_waste_per_meal",
            "predicted_waste_per_100_meals"
        ]
    ].head(20)
)


# ================================================================
# 31. RISK SUMMARY
# ================================================================

risk_summary = (
    decision_support
    .groupby(
        "risk_category",
        observed=False
    )
    .agg(
        observations=(
            "predicted_food_waste_kg",
            "size"
        ),
        average_predicted_waste_kg=(
            "predicted_food_waste_kg",
            "mean"
        ),
        average_actual_waste_kg=(
            "food_waste_kg",
            "mean"
        )
    )
    .reset_index()
)

print("\nRISK SUMMARY")
display(risk_summary)


# ================================================================
# 32. FINAL RESEARCH RESULTS SUMMARY
# ================================================================

print("\n")
print("=" * 80)
print("FINAL RESEARCH RESULTS SUMMARY")
print("=" * 80)

print(
    f"Dataset observations: {len(df)}"
)

print(
    f"Unique dates: {df['date'].nunique()}"
)

print(
    f"Development observations: {len(development_df)}"
)

print(
    f"Final test observations: {len(final_test_df)}"
)

print(
    f"Selected model: {selected_model_name}"
)

print(
    f"Final Test MAE: {final_mae:.4f} kg"
)

print(
    f"Final Test RMSE: {final_rmse:.4f} kg"
)

print(
    f"Final Test R²: {final_r2:.4f}"
)

print(
    f"Selected cluster count: {best_k}"
)

print(
    "Association rules found:",
    len(association_table)
)

print("\nTop SHAP features:")

display(
    shap_importance.head(10)
)


# ================================================================
# 33. EXPORT ALL PAPER TABLES
# ================================================================

OUTPUT_DIR = "paper_results"

os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)

dataset_description.to_csv(
    f"{OUTPUT_DIR}/table_1_dataset_description.csv",
    index=False
)

descriptive_stats.to_csv(
    f"{OUTPUT_DIR}/table_2_descriptive_statistics.csv",
    index=False
)

association_table.to_csv(
    f"{OUTPUT_DIR}/table_3_association_rules.csv",
    index=False
)

model_comparison.to_csv(
    f"{OUTPUT_DIR}/table_4_model_comparison.csv",
    index=False
)

ablation_results_df.to_csv(
    f"{OUTPUT_DIR}/table_5_feature_ablation.csv",
    index=False
)

outlier_results_df.to_csv(
    f"{OUTPUT_DIR}/table_6_outlier_sensitivity.csv",
    index=False
)

final_test_results.to_csv(
    f"{OUTPUT_DIR}/table_7_final_model_test.csv",
    index=False
)

shap_importance.to_csv(
    f"{OUTPUT_DIR}/table_8_shap_feature_importance.csv",
    index=False
)

decision_support.to_csv(
    f"{OUTPUT_DIR}/table_9_decision_support_predictions.csv",
    index=False
)

cluster_profiles.to_csv(
    f"{OUTPUT_DIR}/cluster_profiles.csv",
    index=False
)

cluster_scores_df.to_csv(
    f"{OUTPUT_DIR}/cluster_scores.csv",
    index=False
)

outlier_thresholds_df.to_csv(
    f"{OUTPUT_DIR}/outlier_thresholds.csv",
    index=False
)

prediction_results.to_csv(
    f"{OUTPUT_DIR}/final_predictions.csv",
    index=False
)

monthly_olap.to_csv(
    f"{OUTPUT_DIR}/olap_monthly.csv",
    index=False
)

event_olap.to_csv(
    f"{OUTPUT_DIR}/olap_event.csv",
    index=False
)

staff_olap.to_csv(
    f"{OUTPUT_DIR}/olap_staff.csv",
    index=False
)


# ================================================================
# 34. CREATE ONE EXCEL WORKBOOK FOR THE PAPER
# ================================================================

excel_path = (
    f"{OUTPUT_DIR}/food_waste_research_results.xlsx"
)

with pd.ExcelWriter(
    excel_path,
    engine="openpyxl"
) as writer:

    dataset_description.to_excel(
        writer,
        sheet_name="Table_1_Dataset",
        index=False
    )

    descriptive_stats.to_excel(
        writer,
        sheet_name="Table_2_Statistics",
        index=False
    )

    association_table.to_excel(
        writer,
        sheet_name="Table_3_Association",
        index=False
    )

    model_comparison.to_excel(
        writer,
        sheet_name="Table_4_Model_Comparison",
        index=False
    )

    ablation_results_df.to_excel(
        writer,
        sheet_name="Table_5_Ablation",
        index=False
    )

    outlier_results_df.to_excel(
        writer,
        sheet_name="Table_6_Outliers",
        index=False
    )

    final_test_results.to_excel(
        writer,
        sheet_name="Table_7_Final_Test",
        index=False
    )

    shap_importance.to_excel(
        writer,
        sheet_name="Table_8_SHAP",
        index=False
    )

    decision_support.to_excel(
        writer,
        sheet_name="Table_9_Decision",
        index=False
    )

    cluster_profiles.to_excel(
        writer,
        sheet_name="Cluster_Profiles",
        index=False
    )

    cluster_scores_df.to_excel(
        writer,
        sheet_name="Cluster_Scores",
        index=False
    )

    monthly_olap.to_excel(
        writer,
        sheet_name="OLAP_Monthly",
        index=False
    )

    event_olap.to_excel(
        writer,
        sheet_name="OLAP_Event",
        index=False
    )

    staff_olap.to_excel(
        writer,
        sheet_name="OLAP_Staff",
        index=False
    )


# ================================================================
# 35. FINAL MESSAGE
# ================================================================

print("\n" + "=" * 80)
print("COMPLETE RESEARCH PIPELINE FINISHED")
print("=" * 80)

print(
    f"\nResults saved in: {OUTPUT_DIR}/"
)

print(
    f"Excel workbook: {excel_path}"
)

print("\nPaper tables generated:")

print("1. Dataset Description")
print("2. Descriptive Statistics")
print("3. Association Rules")
print("4. Time-Aware Model Comparison")
print("5. Feature Ablation")
print("6. Outlier Sensitivity")
print("7. Final Model Test Performance")
print("8. SHAP Feature Importance")
print("9. Decision-Support Predictions")

print("\nFigures generated:")
print("1. Food-waste distribution")
print("2. Actual vs predicted")
print("3. Residual analysis")
print("4. SHAP summary")

print("\nFinal model:", selected_model_name)
print(f"Final test MAE: {final_mae:.4f} kg")
print(f"Final test RMSE: {final_rmse:.4f} kg")
print(f"Final test R²: {final_r2:.4f}")

print("\nIMPORTANT FOR THE PAPER:")
print(
    "Do not claim the selected model is universally superior. "
    "State that it was selected according to the predefined "
    "time-aware validation criterion."
)
```
